## Juntando as bases

```python
#Importações
import pandas as pd
import re

from Apoio.config import (
    DADOS_2005_2014,
    DADOS_2015_2019,
    DADOS_2020,
    DADOS_2021,
    DADOS_2022,
    DADOS_2023,
    DADOS_2024,
    DADOS_CONSOLIDADOS,
)
```

#### Concatenando os dataframes

```python
df_2005_2014 = pd.read_csv(DADOS_2005_2014)
df_2015_2019 = pd.read_csv(DADOS_2015_2019)
df_2020 = pd.read_csv(DADOS_2020)
df_2021 = pd.read_csv(DADOS_2021)
df_2022 = pd.read_csv(DADOS_2022)
df_2023 = pd.read_csv(DADOS_2023, encoding="latin")
df_2024 = pd.read_csv(DADOS_2024, encoding="latin")
```

```python
df = pd.concat([df_2005_2014, df_2015_2019, df_2020, df_2021, df_2022, df_2023, df_2024])

df = df.reset_index(drop=True)
```

#### Traduzindo os nomes das colunas

| Original               | Tradução                    |
| ---------------------- | --------------------------- |
| `model year`           | `ano_modelo`                |
| `make`                 | `marca`                     |
| `model`                | `modelo`                    |
| `vehicle class`        | `classe_veiculo`            |
| `engine size (l)`      | `tamanho_motor_l`           |
| `cylinders`            | `cilindros`                 |
| `transmission`         | `transmissao`               |
| `fuel type`            | `tipo_combustivel`          |
| `city (l/100 km)`      | `consumo_cidade_l_100km`    |
| `highway (l/100 km)`   | `consumo_estrada_l_100km`   |
| `combined (l/100 km)`  | `consumo_combinado_l_100km` |
| `combined (mpg)`       | `consumo_combinado_mpg`     |
| `co2 emissions (g/km)` | `emissoes_co2_g_km`         |
| `co2 rating`           | `classificacao_co2`         |
| `smog rating`          | `classificacao_smog`        |

```python
df.columns = [
    'ano_modelo',
    'marca',
    'modelo',
    'classe_veiculo',
    'tamanho_motor_l',
    'cilindros',
    'transmissao',
    'tipo_combustivel',
    'consumo_cidade_l_100km',
    'consumo_estrada_l_100km',
    'consumo_combinado_l_100km',
    'consumo_combinado_mpg',
    'emissoes_co2_g_km',
    'classificacao_co2',
    'classificacao_smog'
]
```
